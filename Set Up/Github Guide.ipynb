## 🔀 Section 1: Introduction to Git & GitHub

### 📚 What is Version Control?

Version control is the process of tracking and managing changes to software code. It's essential for collaborative work and long-term development. Think of it like "Track Changes" in Word, but much more powerful and automated.

#### Benefits:
- **Track progress** over time and understand how your codebase evolved.
- **Collaborate** with others without overwriting each other's work.
- **Revert** to earlier versions if bugs are introduced.
- **Experiment safely** using branches and merging changes when stable.

Every change you make gets logged. If you break something, you can undo it. If someone else adds a new feature, you can review, test, and combine it with your own.

### 📃 Git vs GitHub

Let’s clear up a common confusion: **Git is not GitHub**.

| Feature       | Git                           | GitHub                                  |
| ------------- | ----------------------------- | --------------------------------------- |
| Type          | Command-line tool             | Cloud-based hosting platform            |
| Purpose       | Tracks changes locally        | Stores Git repos online                 |
| Collaboration | Local (can share manually)    | Built-in collaboration tools (PRs)      |
| Use Case      | Work on files, commit changes | Publish repos, collaborate, submit work |

- **Git** runs on your machine. It helps track changes locally and lets you commit updates.
- **GitHub** is an online service that stores your Git repositories in the cloud, so your teammates (and instructors) can see your work.

### 🚀 Why Use Git & GitHub for This Project?

Your term project has multiple moving parts:
- `extract.py`, `transform.py`, `load.py`
- `streamlit_app.py`

Using Git and GitHub ensures:

## 🔧 Section 2: Git Setup & Initialization

Before you can track changes or push code to GitHub, you need to set up Git on your computer and initialize your project repository.

### 🛠️ 1. Install Git

#### For Windows:

1. Go to [https://git-scm.com/download/win](https://git-scm.com/download/win) and download the latest installer.
2. Run the installer. During setup:
   - Choose **"Git from the command line and also from 3rd-party software"**.
   - Choose **"Use Git and optional Unix tools from the Windows Command Prompt"** (this lets you use Git in regular CMD/PowerShell).
   - Leave all other settings at default unless you know what you're doing.
3. After installation, open **Git Bash** from the Start Menu.
4. Run this command to check if Git was installed:
```bash
git --version
```
You should see something like: `git version 2.43.0.windows.1`

#### For macOS (Intel or ARM Macs):

**Option 1: Install via Homebrew (Recommended for both Intel and ARM)**

1. First, check if Homebrew is installed:
```bash
brew --version
```
If you see a version number, you can proceed. If not, install it:
```bash
/bin/bash -c "$(curl -fsSL https://raw.githubusercontent.com/Homebrew/install/HEAD/install.sh)"
```
(If you're on Apple Silicon, follow the on-screen instructions to add Homebrew to your PATH.)

2. Install Git:
```bash
brew install git
```

3. Verify Git installation:
```bash
git --version
```
Example output: `git version 2.42.0`

**Option 2: Use Xcode Command Line Tools (simpler but may be outdated)**
```bash
xcode-select --install
```
This will prompt a system dialog to install command line developer tools.

After installation:
```bash
git --version
```

### 👤 2. Configure Git (One-Time Setup)

You must set your name and email before using Git. This information is attached to each commit you make, so your team and instructors know who contributed what.

#### What to Use:
- Use **your real name** as you'd like it to appear in contribution logs.
- Use the **same email as your GitHub account** to ensure commits link to your GitHub profile.

To find the email you use with GitHub:
1. Go to https://github.com
2. Click your profile picture (top-right) > **Settings**
3. Under **"Emails"**, use the **primary email** listed there

To configure Git, enter these commands in your terminal:
```bash
git config --global user.name "Your Full Name"
git config --global user.email "your@email.com"
```
_(Replace the name and email with yours)_

To verify your settings:
```bash
git config --list
```
You should see something like:
```
user.name=John Doe
user.email=johndoe123@gmail.com
```

🔐 **Note:**
- `--global` applies your settings to all projects. Use this unless you're on a shared computer.
- If you're on a **shared or public machine**, leave out `--global` to set name/email per project:
```bash
git config user.name "Your Name"
git config user.email "your@email.com"
```

Setting this up ensures proper attribution and prevents GitHub from showing commits as "unverified" or "anonymous".

---
### 📂 3. Initialize Git in Your Project Folder

1. Open a terminal 

2. Navigate to your project directory using the `cd` (change directory) command. For example, if your project is inside a folder named `term_project` in your **Documents** folder:
```bash
cd ~/Documents/term_project/
```

3. Initialize a Git repository in this folder:
```bash
git init
```
This creates a hidden `.git` directory which Git uses to store the entire history of changes for this folder. You will not see `.git` unless you use a command like `ls -a` to show hidden files.

4. Confirm everything is set up by checking the Git status:
```bash
git status
```
If Git is working, you should see something like:
```
On branch master

No commits yet

Untracked files:
  (use "git add <file>..." to include in what will be committed)
    extract.py
    load.py
    transform.py
    ...
```
This means Git is ready and watching your files.

---

### ✉️ 4. Create a `.gitignore` File

`.gitignore` tells Git what **not to track**. For example, cache files, credentials, editor configs, etc., that should not go into your repo.

Common entries for Python data projects:
```
__pycache__/
*.pyc
.env
.vscode/
.idea/
.DS_Store
*.db
```
To create a `.gitignore` file:
```bash
touch .gitignore
```

Paste the contents listed above into it using any text editor (e.g., Notepad, VS Code, nano).

Then verify Git is ignoring the right files:
```bash
git status
```
You should **not** see `.env`, `.pyc` files, or `.vscode/` in the untracked files list.

---

### 📅 5. First Commit

A commit is a snapshot of your current codebase.

1. Add your files to the Git staging area:
```bash
git add .
```
- The dot `.` means "stage everything in this folder."
- If you want to add individual files, do:
```bash
git add extract.py transform.py
```

2. Make your first commit:
```bash
git commit -m "Initial commit: setup term project"
```
Use meaningful commit messages that describe what you did. Examples:
- `Added Streamlit app and dashboard layout`
- `Updated transform.py to handle nulls`

You can repeat the add + commit process every time you make changes:
```bash
git add .
git commit -m "Describe what you changed"
```

You now have a local Git repository that tracks your work.

In the next section, we'll create a GitHub repository and connect it to your local repo so your group can collaborate and submit your work online.




## 🛜 Section 3: Create and Connect to a GitHub Repository

Once your local Git repository is initialized and has at least one commit, you can upload it to GitHub. This enables collaboration, online backups, and project submission.

---

### 🔢 Step 1: Create a GitHub Repository (Updated)

1. Open [https://github.com](https://github.com) and log in.  
   *Don’t have an account? [Sign up here](https://github.com/signup).*

2. Click the **"+"** icon at the top-right corner → **New repository**.

3. Set the repository details:
   - **Repository Name**: Use something descriptive like `term_project_group7`.
   - **Visibility**: Choose **Private** if you’re working in a group and don’t want it public yet. You can change this later.
   - ⚠️ **Important**: **Do NOT initialize** the repository with a README, `.gitignore`, or license — your local project already contains these files.

4. Click **Create repository**.

After this, you will land on a page with GitHub CLI instructions. Keep this tab open.

Note: Click on **Code** and copy the url which can be used in the next step.

---

### 🔗 Step 2: Link Local Repository to GitHub
In your terminal (inside your project folder):

1. Add GitHub remote:
```bash
git remote add origin https://github.com/YourUsername/your-repo-name.git
```
Replace placeholders with your real GitHub username and repository name.

2. Rename your branch to `main` (GitHub default):
```bash
git branch -M main
```

3. Push your local repository to GitHub:
```bash
git push -u origin main
```
---
### 📄 Step 3: Confirm Online

1. Go to: `https://github.com/YourUsername/your-repo-name`
2. You should see:
   - All your files (`extract.py`, `.env.sample`, `README.md`, etc.)
   - Latest commit under "Commits" tab
   - Project tree organized the same way as your local folder

To push future changes:
```bash
git add .
git commit -m "Explain your change"
git push
```

✅ Now your local Git repository is connected to GitHub — you're set for teamwork, submissions, and a real-world workflow.

### 🔐 Optional Authentication: Personal Access Token (PAT)
GitHub no longer supports password authentication for HTTPS pushes.

When prompted for username and password:
- **Username**: Your GitHub username
- **Password**: Use a **personal access token (PAT)**, not your real password

#### 🔍 How to generate a PAT:
1. Go to: [https://github.com/settings/tokens](https://github.com/settings/tokens)
2. Click **Generate new token (classic)**
3. Name your token (e.g., "Project Uploads")
4. Set expiration (30 days is fine for a class project)
5. Check the `repo` scope box
6. Click **Generate token**
7. Copy the token and save it somewhere safe (you won't see it again)

💡 Paste this token when prompted for a password during `git push`.

#### 🧠 Optional: Save Credentials
If you don’t want to retype your token every time:
- On Windows: Git Credential Manager comes built-in with Git for Windows
- On macOS: Use Keychain
- On Linux: Use `git-credential-store` or `git-credential-cache`

---

### 🔌 Optional Setup: SSH Authentication (Advanced)
To avoid using tokens or typing anything at all when pushing:
- Generate SSH keys using: `ssh-keygen -t ed25519 -C "youremail@example.com"`
- Add your public key to GitHub (under Settings → SSH and GPG keys)
- More info: [GitHub SSH Guide](https://docs.github.com/en/authentication/connecting-to-github-with-ssh)

## 🤝 Section 7: Collaborating with Your Team and Keeping the Repo Updated

Once your project is on GitHub, you and your teammates can all work on the same repository without conflict — as long as you follow basic Git collaboration workflows.

#### 🧭 Workflow Overview:

Every team member should:

1. **Clone** the repository
2. **Create a new branch** for the feature they’re working on
3. **Commit** changes locally
4. **Push** the branch to GitHub
5. Create a **Pull Request (PR)** and have it reviewed before merging into `main`

---

#### 🔁 Step 1: Clone the Repository (Only Once)

If someone else created the GitHub repo, others must clone it to their machines:

```bash
git clone https://github.com/YourTeam/term_project.git
```

Then:

```bash
cd term_project
```

---

#### 🌱 Step 2: Create a Branch

Use a new branch for each task or feature:

```bash
git checkout -b add-streamlit-dashboard
```

This creates and switches to a new branch named `add-streamlit-dashboard`

When you’re done working on that feature, commit and push your changes:

```bash
git add .
git commit -m "Add initial Streamlit dashboard"
git push -u origin add-streamlit-dashboard
```

---

#### 🔀 Step 3: Make a Pull Request (PR)

1. Go to your repo on GitHub.
2. Click the “Compare & pull request” button.
3. Add a short title + description (explain what your PR does).
4. Click **Create pull request**.

Once reviewed and approved, you (or someone else) can **Merge** it into the `main` branch.

---

#### 🛠️ Step 4: Keeping Your Local Repo in Sync

Always update your local copy before working:

```bash
git checkout main

git pull origin main
```

To delete old branches after merging:

```bash
git branch -d add-streamlit-dashboard        # deletes locally

git push origin --delete add-streamlit-dashboard   # deletes on GitHub
```

---

### 📌 Best Practices for Team Projects:

- **One branch per feature or bugfix** — never edit `main` directly
- **Make frequent, small commits** with clear messages
- **Pull latest code** before starting any work session
- **Communicate** what branch you're working on to avoid conflicts
- **Use Pull Requests** for all merges — even if you’re the only one working on it
- **Never push ****`.env`**** files or secrets** — use `.gitignore`

By following these practices, you’ll ensure smooth collaboration and reduce bugs or conflicts as you build your project together.
