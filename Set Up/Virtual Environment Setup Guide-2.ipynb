## 🛠️ Step 2.3: Create the Virtual Environment

Now that you’re in the correct working directory and have your custom Python path ready, let’s create the virtual environment for this course.

### 🧪 What You’re Doing
You’ll be creating a folder named `daen328_env` that contains an isolated Python setup. This keeps your course packages separate from other projects or system-wide Python installs.

---

### ⚙️ Instructions
1. Paste your full Python path from Step 1.1 into the `custom_python_path` variable in the next code cell.
2. Run the code cell to create the environment.
3. If the environment already exists, it will let you know.

Once completed, you’ll see confirmation that the folder `daen328_env` has been created in your current working directory.

---


````python
# ✅ Step 2.3: Create the Virtual Environment Using Your Custom Python Path

import os
import subprocess
import sys

# Name of the virtual environment
env_name = "daen328_env"

# Resolve absolute path and validate
custom_python_path = os.path.abspath(os.path.expanduser(custom_python_path))
print(f"🔍 Verifying Python path: {custom_python_path}")

if not os.path.isfile(custom_python_path):
    raise SystemExit("❌ The Python path you provided is not valid. Please check the file path and try again.")

# Check Python version is 3.10 or later
try:
    version_check = subprocess.run([custom_python_path, "--version"], capture_output=True, text=True, check=True)
    version_str = version_check.stdout.strip() or version_check.stderr.strip()
    print(f"📦 Python version found: {version_str}")
    version_parts = version_str.split()[1].split('.')
    major, minor = int(version_parts[0]), int(version_parts[1])
    if major != 3 or minor < 10:
        raise SystemExit("❌ Python version must be 3.10 or higher for this course.")
except Exception as e:
    raise SystemExit(f"❌ Failed to verify Python version: {e}")

# Check if virtual environment already exists
if not os.path.exists(env_name):
    print(f"🔧 Creating virtual environment '{env_name}' using interpreter: {custom_python_path}\n")
    try:
        subprocess.run([custom_python_path, "-m", "venv", env_name], check=True)
        print(f"✅ Virtual environment '{env_name}' created successfully in {os.getcwd()}.")
    except subprocess.CalledProcessError as e:
        raise SystemExit(f"❌ Failed to create virtual environment:\n{e.stderr or str(e)}")
else:
    print(f"⚠️ Virtual environment '{env_name}' already exists in {os.getcwd()}.")
````

### 📦 What Happens Here
1. A new folder named `daen328_env` is created in your current working directory.
2. Inside this folder, you’ll find:
   - A clean, local Python interpreter
   - `pip`, the Python package installer
   - A `Scripts/` or `bin/` folder depending on your OS.
   - All files needed to manage and isolate your environment


### 💡 Tips

## 🧠 Step 3: Register the Virtual Environment in Jupyter

To use your virtual environment inside **Jupyter Notebook**, you need to register it as a new kernel. This allows you to select it from the kernel dropdown in your notebooks.

---

### 📌 What This Step Does
- Installs `ipykernel` into your virtual environment
- Registers your environment as a kernel named **`Python (daen328_env)`**
- Makes it easy to switch between environments directly in Jupyter

### 🪄 What We’re Doing:
1. Activate your virtual environment
2. Install the `ipykernel` package
3. Register your environment as a new Jupyter kernel


> ✅ Once this is done, you'll be able to select "Python (daen328_env)" from the Jupyter kernel menu at the top right of any notebook.


### For Windows

````python
# ✅ Step 3: Install ipykernel and Register the Environment for Windows

import os
import subprocess
import sys

env_name = "daen328_env"
kernel_display_name = f"Python ({env_name})"
# Full path to the virtual environment's Python interpreter (Windows)
venv_python = os.path.join(env_name, "Scripts", "python.exe")

if not os.path.exists(venv_python):
    raise SystemExit(f"❌ Python interpreter not found at {venv_python}. Make sure Step 2 completed successfully.")

# Install ipykernel inside the venv
print("📦 Installing ipykernel in the virtual environment...")
try:
    subprocess.run([venv_python, "-m", "pip", "install", "--quiet", "ipykernel"], check=True)
except subprocess.CalledProcessError as e:
    raise SystemExit(f"❌ Failed to install ipykernel: {e.stderr or str(e)}")

# Register the venv as a Jupyter kernel
print("🔗 Registering the virtual environment as a Jupyter kernel...")
try:
    subprocess.run([
        venv_python, "-m", "ipykernel", "install",
        "--user", "--name", env_name,
        "--display-name", kernel_display_name
    ], check=True)
    print(f"✅ Kernel '{kernel_display_name}' registered successfully!")
except subprocess.CalledProcessError as e:
    raise SystemExit(f"❌ Failed to register kernel: {e.stderr or str(e)}")
````

### For macOS/Linux

````python
# ✅ Step 3: Install ipykernel and Register the Environment for macOS

import os
import subprocess
import sys

env_name = "daen328_env"
kernel_display_name = f"Python ({env_name})"

# Full path to the virtual environment's Python interpreter (macOS/Linux)
venv_python = os.path.join(env_name, "bin", "python3")

if not os.path.exists(venv_python):
    raise SystemExit(f"❌ Python interpreter not found at {venv_python}. Make sure Step 2 completed successfully.")

# Install ipykernel inside the venv
print("📦 Installing ipykernel in the virtual environment...")
try:
    subprocess.run([venv_python, "-m", "pip", "install", "--quiet", "ipykernel"], check=True)
except subprocess.CalledProcessError as e:
    raise SystemExit(f"❌ Failed to install ipykernel: {e.stderr or str(e)}")

# Register the venv as a Jupyter kernel
print("🔗 Registering the virtual environment as a Jupyter kernel...")
try:
    subprocess.run([
        venv_python, "-m", "ipykernel", "install",
        "--user", "--name", env_name,
        "--display-name", kernel_display_name
    ], check=True)
    print(f"✅ Kernel '{kernel_display_name}' registered successfully!")
except subprocess.CalledProcessError as e:
    raise SystemExit(f"❌ Failed to register kernel: {e.stderr or str(e)}")
````

## 🧪 Step 4: Test the Virtual Environment

Now that your virtual environment is registered with Jupyter, let’s verify that it’s working correctly.

````python
# ✅ Step 4: Verify Kernel Connection

import sys
import os

print("✅ Kernel is connected and running.")
print(f"Python Executable Path: {sys.executable}\n")

expected_env_name = "daen328_env"

# Normalize the path for robust checking
normalized_path = os.path.normpath(sys.executable).lower()

if expected_env_name.lower() in normalized_path:
    print(f"🎉 The kernel is using the correct virtual environment: '{expected_env_name}'")
else:
    print("❌ The kernel is not linked to the expected virtual environment.")
    print("🔁 Please go to the Kernel menu above and select 'Python (daen328_env)'.")
    print("🛠️ If the kernel doesn't appear, restart Jupyter Notebook and try again.")
````

### 🛠️ Troubleshooting
If the environment doesn't seem to be active:

````python
# ✅ Step 5: Verify and Upgrade pip in the Virtual Environment

import sys
import subprocess

# Step 5.1: Check pip version
print("🔍 Checking pip version inside the virtual environment...\n")
try:
    result = subprocess.run([sys.executable, "-m", "pip", "--version"], capture_output=True, text=True, check=True)
    print(f"✅ pip Version: {result.stdout.strip()}\n")
except subprocess.CalledProcessError as e:
    print("❌ Unable to check pip version. Please ensure the virtual environment is active.\n")
    sys.exit(1)

# Step 5.2: Upgrade pip
print("📦 Upgrading pip to the latest version...\n")
try:
    upgrade_result = subprocess.run([sys.executable, "-m", "pip", "install", "--upgrade", "pip"], capture_output=True, text=True, check=True)
    print(f"✅ pip has been upgraded successfully:\n{upgrade_result.stdout.strip()}\n")
except subprocess.CalledProcessError as e:
    print(f"❌ pip upgrade failed. Error:\n{e.stderr.strip() if e.stderr else str(e)}\n")
````
